```python
# 2. 100% Disk-Backed Kaggle SSD Setup (~20s!): Blender 4.2.3 + MPFB2 + Authentic 17th-Century V8 4K Cast Assets in /kaggle/working/.cache
import os, subprocess, time, gc
from concurrent.futures import ThreadPoolExecutor

t0 = time.time()
CACHE_DIR = '/kaggle/working/.cache'
BLENDER_BIN = f'{CACHE_DIR}/blender/blender'
PACKS_DIR = f'{CACHE_DIR}/mpfb_packs'
ASSET_MARKER = f'{CACHE_DIR}/.mpfb_newton_v8_assets_installed'
os.makedirs(PACKS_DIR, exist_ok=True)

# Clean any old files from /tmp so 100% of Kaggle System RAM stays completely free!
subprocess.run('rm -rf /tmp/blender* /tmp/mpfb* /tmp/gpu*.log', shell=True)

FAST_ASSET_URL = 'https://raw.githubusercontent.com/ssghtc002-sudo/isaac-newton-gravity-prism-3d/assets-v8/mpfb_newton_v8_4k_assets.zip'
FAST_ASSET_DST = f'{PACKS_DIR}/mpfb_newton_v8_4k_assets.zip'

def setup_blender():
    if os.path.exists(BLENDER_BIN):
        return 'Blender 4.2.3 cached on /kaggle/working SSD'
    t_b = time.time()
    subprocess.run(
        f'curl -fSL --retry 3 https://download.blender.org/release/Blender4.2/blender-4.2.3-linux-x64.tar.xz -o {CACHE_DIR}/blender.tar.xz '
        f'&& mkdir -p {CACHE_DIR}/blender && tar -xf {CACHE_DIR}/blender.tar.xz -C {CACHE_DIR}/blender --strip-components=1 '
        f'&& rm -f {CACHE_DIR}/blender.tar.xz',
        shell=True, check=True
    )
    return f'Blender 4.2.3 installed on SSD ({time.time()-t_b:.1f}s)'

def setup_mpfb_zip():
    if os.path.exists(f'{CACHE_DIR}/.mpfb_ext_installed') or os.path.exists(f'{CACHE_DIR}/mpfb.zip'):
        return 'MPFB2 cached'
    t_m = time.time()
    subprocess.run(
        f'rm -rf {CACHE_DIR}/mpfb2 && git clone --depth 1 -q https://github.com/makehumancommunity/mpfb2.git {CACHE_DIR}/mpfb2 '
        f'&& (cd {CACHE_DIR}/mpfb2/src && zip -qr {CACHE_DIR}/mpfb.zip mpfb) && rm -rf {CACHE_DIR}/mpfb2',
        shell=True, check=True
    )
    return f'MPFB2 zip built ({time.time()-t_m:.1f}s)'

def setup_4k_assets():
    if os.path.exists(ASSET_MARKER):
        return '4K Authentic 17th-Century V8 Newton & Halley Cast Assets already extracted'
    if os.path.exists(FAST_ASSET_DST) and os.path.getsize(FAST_ASSET_DST) > 25_000_000:
        return f'4K Cast Assets zip cached ({os.path.getsize(FAST_ASSET_DST)/1e6:.1f} MB)'
    t_a = time.time()
    subprocess.run(
        f'curl -fSL --retry 3 "{FAST_ASSET_URL}" -o "{FAST_ASSET_DST}"',
        shell=True, check=True
    )
    return f'4K V8 Cast Assets downloaded from GitHub CDN ({os.path.getsize(FAST_ASSET_DST)/1e6:.2f} MB in {time.time()-t_a:.2f}s!)'

with ThreadPoolExecutor(max_workers=3) as ex:
    for f in [ex.submit(setup_blender), ex.submit(setup_mpfb_zip), ex.submit(setup_4k_assets)]:
        print('[OK]', f.result())

if not os.path.exists(f'{CACHE_DIR}/.mpfb_ext_installed'):
    !{BLENDER_BIN} -b --command extension install-file -r user_default -e {CACHE_DIR}/mpfb.zip
    open(f'{CACHE_DIR}/.mpfb_ext_installed', 'w').write('ok')
    subprocess.run(f'rm -f {CACHE_DIR}/mpfb.zip', shell=True)

if not os.path.exists(ASSET_MARKER):
    !MPFB_PACKS_DIR={PACKS_DIR} {BLENDER_BIN} --background --python src/install_mpfb_packs.py
    open(ASSET_MARKER, 'w').write('ok')
    subprocess.run(f'rm -rf {PACKS_DIR}', shell=True)

gc.collect()
!free -h && df -h /kaggle/working
print(f'=== 100% KAGGLE SSD V9 SETUP COMPLETE in {time.time()-t0:.1f}s! ===')
```

```python
# 4. 1-Click Ultra-Turbo Dual-T4 Render + True Full HD (1080x1920) ARRI-Alexa MP4 Export + FileLink Download!
import os, shutil, subprocess, time, glob, gc
from IPython.display import FileLink, Video, display

# Auto-Bootstrap Guard: if Kaggle restarted the VM (e.g. after toggling GPU/Internet), auto-clone repo & setup SSD cache!
REPO_DIR = '/kaggle/working/isaac-newton-gravity-prism-3d'
CACHE_DIR = '/kaggle/working/.cache'
BLENDER_BIN = f'{CACHE_DIR}/blender/blender'
ASSET_MARKER = f'{CACHE_DIR}/.mpfb_newton_v8_assets_installed'
if not os.path.isdir(REPO_DIR):
    print('Repo missing after Kaggle VM reset -> cloning fresh repo into /kaggle/working...')
    subprocess.run(['git', 'clone', '--depth', '1', 'https://github.com/ssghtc002-sudo/isaac-newton-gravity-prism-3d.git', REPO_DIR], check=True)
os.chdir(REPO_DIR)
if not os.path.exists(BLENDER_BIN) or not os.path.exists(ASSET_MARKER):
    print('SSD cache missing after Kaggle VM reset -> auto-installing Blender 4.2.3 + MPFB2 + V8 4K Cast (~20s)...')
    PACKS_DIR = f'{CACHE_DIR}/mpfb_packs'
    os.makedirs(PACKS_DIR, exist_ok=True)
    if not os.path.exists(BLENDER_BIN):
        subprocess.run(f'curl -fSL --retry 3 https://download.blender.org/release/Blender4.2/blender-4.2.3-linux-x64.tar.xz -o {CACHE_DIR}/blender.tar.xz && mkdir -p {CACHE_DIR}/blender && tar -xf {CACHE_DIR}/blender.tar.xz -C {CACHE_DIR}/blender --strip-components=1 && rm -f {CACHE_DIR}/blender.tar.xz', shell=True, check=True)
    if not os.path.exists(f'{CACHE_DIR}/.mpfb_ext_installed'):
        subprocess.run(f'rm -rf {CACHE_DIR}/mpfb2 && git clone --depth 1 -q https://github.com/makehumancommunity/mpfb2.git {CACHE_DIR}/mpfb2 && (cd {CACHE_DIR}/mpfb2/src && zip -qr {CACHE_DIR}/mpfb.zip mpfb) && rm -rf {CACHE_DIR}/mpfb2', shell=True, check=True)
        subprocess.run(f'{BLENDER_BIN} -b --command extension install-file -r user_default -e {CACHE_DIR}/mpfb.zip', shell=True, check=True)
        open(f'{CACHE_DIR}/.mpfb_ext_installed', 'w').write('ok')
        subprocess.run(f'rm -f {CACHE_DIR}/mpfb.zip', shell=True)
    if not os.path.exists(ASSET_MARKER):
        subprocess.run(f'curl -fSL --retry 3 https://raw.githubusercontent.com/ssghtc002-sudo/isaac-newton-gravity-prism-3d/assets-v8/mpfb_newton_v8_4k_assets.zip -o {PACKS_DIR}/mpfb_newton_v8_4k_assets.zip', shell=True, check=True)
        subprocess.run(f'MPFB_PACKS_DIR={PACKS_DIR} {BLENDER_BIN} --background --python src/install_mpfb_packs.py', shell=True, check=True)
        open(ASSET_MARKER, 'w').write('ok')
        subprocess.run(f'rm -rf {PACKS_DIR}', shell=True)
gc.collect()
subprocess.run('pkill -9 -f "blender/blender" || true', shell=True)

LOG_DIR = '/kaggle/working/.cache'
BAKED_BLEND = f'{LOG_DIR}/newton_v9_baked.blend'
BAKE_LOG = f'{LOG_DIR}/bake.log'
FINAL_MP4 = '/kaggle/working/isaac_newton_gravity_prism_short.mp4'
TOTAL = 720
shutil.rmtree('render_frames', ignore_errors=True)

# Step A: Bake 720-frame scene ONCE into SSD cache (logged cleanly to bake.log -> zero WebSocket spam!)
src_mtime = max(os.path.getmtime(p) for p in glob.glob('src/*.py'))
if not os.path.exists(BAKED_BLEND) or os.path.getmtime(BAKED_BLEND) < src_mtime:
    t_bk = time.time()
    print('Baking 720-frame V9 scene once into SSD cache (~25-35s, zero duplicate worker CPU build!)...', flush=True)
    bk = subprocess.run(
        f'OMP_NUM_THREADS=2 {BLENDER_BIN} --background --python-exit-code 1 --python src/build_newton_scene.py -- --no-render --save-blend {BAKED_BLEND} --fps 24 --samples 12 --scale 75',
        shell=True, stdout=open(BAKE_LOG, 'w'), stderr=subprocess.STDOUT
    )
    if bk.returncode != 0:
        print('--- BAKE LOG TAIL ---\n' + ''.join(open(BAKE_LOG).readlines()[-35:]))
        raise RuntimeError(f'Scene bake failed (code {bk.returncode})')
    print(f'Scene baked ({os.path.getsize(BAKED_BLEND)/1e6:.1f} MB) in {time.time()-t_bk:.1f}s!', flush=True)

# Step B: Dual-T4 OPTIX + Fast GPU OIDN Render (720 frames @ 24fps, 12 spp)
gpu_count = int(subprocess.check_output('nvidia-smi --list-gpus | wc -l', shell=True).decode().strip())
print(f'Detected {gpu_count} Kaggle NVIDIA T4 GPU(s)! Starting V9 OPTIX Ultra-Turbo render ({TOTAL} frames @ 24fps, 12 spp -> Full HD 1080x1920)...', flush=True)
t0 = time.time()

if gpu_count >= 2:
    p0 = subprocess.Popen(
        f'CUDA_VISIBLE_DEVICES=0 OMP_NUM_THREADS=2 {BLENDER_BIN} --background {BAKED_BLEND} --python-exit-code 1 --python src/render_worker.py -- --start-frame 1 --end-frame 360 --fps 24 --samples 12 --scale 75',
        shell=True, stdout=open(f'{LOG_DIR}/gpu0.log', 'w'), stderr=subprocess.STDOUT
    )
    time.sleep(2)
    p1 = subprocess.Popen(
        f'CUDA_VISIBLE_DEVICES=1 OMP_NUM_THREADS=2 {BLENDER_BIN} --background {BAKED_BLEND} --python-exit-code 1 --python src/render_worker.py -- --start-frame 361 --end-frame 720 --fps 24 --samples 12 --scale 75',
        shell=True, stdout=open(f'{LOG_DIR}/gpu1.log', 'w'), stderr=subprocess.STDOUT
    )
    while p0.poll() is None or p1.poll() is None:
        done = len(glob.glob('render_frames/frame_*.jpg'))
        elapsed = time.time() - t0
        rate = done / max(1.0, elapsed)
        eta = (TOTAL - done) / max(0.01, rate) if done > 2 else 0
        print(f'\r[Kaggle Dual-T4 V9 Ultra-Turbo] {done}/{TOTAL} frames ({done*100/TOTAL:.1f}%) | Elapsed: {elapsed/60:.1f}m | ETA: {eta/60:.1f}m   ', end='', flush=True)
        time.sleep(3)
    print()
    if p0.returncode != 0 or p1.returncode != 0:
        print('--- GPU 0 LOG TAIL ---\n' + ''.join(open(f'{LOG_DIR}/gpu0.log').readlines()[-25:]))
        print('--- GPU 1 LOG TAIL ---\n' + ''.join(open(f'{LOG_DIR}/gpu1.log').readlines()[-25:]))
        raise RuntimeError(f'Worker error: p0={p0.returncode} p1={p1.returncode}')
else:
    subprocess.run(
        f'OMP_NUM_THREADS=2 {BLENDER_BIN} --background {BAKED_BLEND} --python-exit-code 1 --python src/render_worker.py -- --start-frame 1 --end-frame 720 --fps 24 --samples 12 --scale 75',
        shell=True, stdout=open(f'{LOG_DIR}/gpu0.log', 'w'), stderr=subprocess.STDOUT, check=True
    )

rendered = len(glob.glob('render_frames/frame_*.jpg'))
assert rendered == TOTAL, f'Expected {TOTAL} rendered frames, found {rendered}!'
print(f'Render complete: {rendered}/{TOTAL} frames in {(time.time()-t0)/60:.1f} min! Encoding Full HD 1080x1920 MP4...', flush=True)

# Step C: Final Full HD (1080x1920) ARRI-Alexa Filmic Finishing + Mastered Audio + Karaoke Subtitles
t_enc = time.time()
vf_graph = (
    '[0:v]scale=1080:1920:flags=lanczos:accurate_rnd=1,unsharp=3:3:0.35:3:3:0.0,format=gbrp,split[base][hi];'
    '[hi]eq=brightness=-0.42:contrast=2.4,gblur=sigma=16,format=gbrp[glow];'
    '[base][glow]blend=all_mode=screen:all_opacity=0.28,'
    'colorbalance=rs=-0.012:bs=0.020:rh=0.022:bh=-0.016,'
    'vignette=PI/6.5:eval=init,format=yuv420p,noise=alls=3:allf=t,'
    'ass=subtitles/subtitles.ass:fontsdir=subtitles/fonts[vout]'
)
subprocess.run([
    'ffmpeg', '-y', '-framerate', '24', '-i', 'render_frames/frame_%04d.jpg',
    '-i', 'audio/output/master_audio.wav',
    '-filter_complex', vf_graph,
    '-map', '[vout]', '-map', '1:a',
    '-c:v', 'libx264', '-preset', 'medium', '-profile:v', 'high', '-crf', '15', '-pix_fmt', 'yuv420p',
    '-c:a', 'aac', '-b:a', '192k', '-shortest', '-movflags', '+faststart',
    FINAL_MP4, '-loglevel', 'warning'
], check=True)
print(f'Full HD 1080x1920 MP4 encoded in {time.time()-t_enc:.1f}s!')

# Step D: Clean intermediate files so /kaggle/working stays spotless for Kaggle Output Panel!
shutil.rmtree('render_frames', ignore_errors=True)
shutil.rmtree('qa_frames', ignore_errors=True)
if os.path.exists(BAKED_BLEND):
    os.remove(BAKED_BLEND)

print('=== VERIFIED FULL HD (1080x1920) EXPORT STREAM INFO ===')
print(subprocess.check_output([
    'ffprobe', '-v', 'error', '-select_streams', 'v:0',
    '-show_entries', 'stream=width,height,r_frame_rate,nb_frames,duration',
    '-of', 'compact', FINAL_MP4
]).decode().strip())
print(f'File: {FINAL_MP4} ({os.path.getsize(FINAL_MP4)/1e6:.2f} MB) | Total End-to-End Time: {(time.time()-t0)/60:.1f} min!')

os.chdir('/kaggle/working')
print('\n=== TAP LINK BELOW OR OPEN KAGGLE RIGHT-SIDE OUTPUT PANEL TO DOWNLOAD FULL HD 1080x1920 MP4 ===')
display(FileLink('isaac_newton_gravity_prism_short.mp4'))
display(Video(FINAL_MP4, embed=False, width=320))
```
