```python
# 2. 100% Disk-Backed Kaggle SSD Setup (~20s!): Blender 4.2.3 + MPFB2 + 43.5 MB Authentic 17th-Century V7 Cast Assets in /kaggle/working/.cache
import os, subprocess, time, gc
from concurrent.futures import ThreadPoolExecutor

t0 = time.time()
CACHE_DIR = '/kaggle/working/.cache'
BLENDER_BIN = f'{CACHE_DIR}/blender/blender'
PACKS_DIR = f'{CACHE_DIR}/mpfb_packs'
ASSET_MARKER = f'{CACHE_DIR}/.mpfb_newton_v7_assets_installed'
os.makedirs(PACKS_DIR, exist_ok=True)

# Clean any old files from /tmp so 100% of Kaggle System RAM stays completely free!
subprocess.run('rm -rf /tmp/blender* /tmp/mpfb* /tmp/gpu*.log', shell=True)

FAST_ASSET_URL = 'https://raw.githubusercontent.com/ssghtc002-sudo/isaac-newton-gravity-prism-3d/assets-v7/mpfb_newton_v7_4k_assets.zip'
FAST_ASSET_DST = f'{PACKS_DIR}/mpfb_newton_v7_4k_assets.zip'

def setup_blender():
    if os.path.exists(BLENDER_BIN):
        return 'Blender 4.2.3 cached on /kaggle/working SSD'
    t_b = time.time()
    subprocess.run(
        f'curl -fSL --retry 3 https://download.blender.org/release/Blender4.2/blender-4.2.3-linux-x64.tar.xz -o {CACHE_DIR}/blender.tar.xz '
        f'&& mkdir -p {CACHE_DIR}/blender && tar -xf {CACHE_DIR}/blender.tar.xz -C {CACHE_DIR}/blender --strip-components=1 '
        f'&& rm -f {CACHE_DIR}/blender.tar.xz',
        shell=True, check=True
    )
    return f'Blender 4.2.3 installed on SSD ({time.time()-t_b:.1f}s)'

def setup_mpfb_zip():
    if os.path.exists(f'{CACHE_DIR}/.mpfb_ext_installed') or os.path.exists(f'{CACHE_DIR}/mpfb.zip'):
        return 'MPFB2 cached'
    t_m = time.time()
    subprocess.run(
        f'rm -rf {CACHE_DIR}/mpfb2 && git clone --depth 1 -q https://github.com/makehumancommunity/mpfb2.git {CACHE_DIR}/mpfb2 '
        f'&& (cd {CACHE_DIR}/mpfb2/src && zip -qr {CACHE_DIR}/mpfb.zip mpfb) && rm -rf {CACHE_DIR}/mpfb2',
        shell=True, check=True
    )
    return f'MPFB2 zip built ({time.time()-t_m:.1f}s)'

def setup_4k_assets():
    if os.path.exists(ASSET_MARKER):
        return '4K Authentic 17th-Century V7 Newton & Halley Cast Assets already extracted'
    if os.path.exists(FAST_ASSET_DST) and os.path.getsize(FAST_ASSET_DST) > 35_000_000:
        return f'4K Cast Assets zip cached ({os.path.getsize(FAST_ASSET_DST)/1e6:.1f} MB)'
    t_a = time.time()
    subprocess.run(
        f'curl -fSL --retry 3 "{FAST_ASSET_URL}" -o "{FAST_ASSET_DST}"',
        shell=True, check=True
    )
    return f'4K V7 Cast Assets downloaded from GitHub CDN ({os.path.getsize(FAST_ASSET_DST)/1e6:.2f} MB in {time.time()-t_a:.2f}s!)'

with ThreadPoolExecutor(max_workers=3) as ex:
    for f in [ex.submit(setup_blender), ex.submit(setup_mpfb_zip), ex.submit(setup_4k_assets)]:
        print('[OK]', f.result())

if not os.path.exists(f'{CACHE_DIR}/.mpfb_ext_installed'):
    !{BLENDER_BIN} -b --command extension install-file -r user_default -e {CACHE_DIR}/mpfb.zip
    open(f'{CACHE_DIR}/.mpfb_ext_installed', 'w').write('ok')
    subprocess.run(f'rm -f {CACHE_DIR}/mpfb.zip', shell=True)

if not os.path.exists(ASSET_MARKER):
    !MPFB_PACKS_DIR={PACKS_DIR} {BLENDER_BIN} --background --python src/install_mpfb_packs.py
    open(ASSET_MARKER, 'w').write('ok')
    subprocess.run(f'rm -rf {PACKS_DIR}', shell=True)

gc.collect()
!free -h && df -h /kaggle/working
print(f'=== 100% KAGGLE SSD V7 SETUP COMPLETE in {time.time()-t0:.1f}s! ===')
```

```python
# 4. Zero-Quality-Drop Kaggle Dual-T4 OPTIX Turbo Render (720 frames @ 24fps, 12 spp OPTIX RGB_ALBEDO_NORMAL, Full 75%->1080p, ~6.5-8 min!)
%cd /kaggle/working/isaac-newton-gravity-prism-3d
import subprocess, time, glob, gc

gc.collect()
subprocess.run('pkill -9 -f "blender/blender" || true', shell=True)

BLENDER_BIN = '/kaggle/working/.cache/blender/blender'
LOG_DIR = '/kaggle/working/.cache'
TOTAL = 720
!rm -rf render_frames
gpu_count = int(subprocess.check_output('nvidia-smi --list-gpus | wc -l', shell=True).decode().strip())
print(f'Detected {gpu_count} Kaggle NVIDIA T4 GPU(s)! Starting Zero-Quality-Drop V7 OPTIX Turbo render ({TOTAL} frames @ 24fps, 12 spp)...')
t0 = time.time()

if gpu_count >= 2:
    p0 = subprocess.Popen(
        f'CUDA_VISIBLE_DEVICES=0 OMP_NUM_THREADS=1 {BLENDER_BIN} --background --python src/build_newton_scene.py -- --start-frame 1 --end-frame 360 --fps 24 --samples 12 --scale 75',
        shell=True, stdout=open(f'{LOG_DIR}/gpu0.log', 'w'), stderr=subprocess.STDOUT
    )
    # Stagger GPU 1 by 10s so both workers never peak CPU RAM during MPFB2 4K scene build simultaneously!
    time.sleep(10)
    p1 = subprocess.Popen(
        f'CUDA_VISIBLE_DEVICES=1 OMP_NUM_THREADS=1 {BLENDER_BIN} --background --python src/build_newton_scene.py -- --start-frame 361 --end-frame 720 --fps 24 --samples 12 --scale 75',
        shell=True, stdout=open(f'{LOG_DIR}/gpu1.log', 'w'), stderr=subprocess.STDOUT
    )
    while p0.poll() is None or p1.poll() is None:
        done = len(glob.glob('render_frames/frame_*.jpg'))
        elapsed = time.time() - t0
        rate = done / max(1.0, elapsed)
        eta = (TOTAL - done) / max(0.01, rate) if done > 2 else 0
        print(f'\r[Kaggle Dual-T4 OPTIX Turbo] {done}/{TOTAL} frames ({done*100/TOTAL:.1f}%) | Elapsed: {elapsed/60:.1f}m | ETA: {eta/60:.1f}m   ', end='', flush=True)
        time.sleep(3)
    print()
    if p0.returncode != 0 or p1.returncode != 0:
        print('--- GPU 0 LOG TAIL ---\n' + ''.join(open(f'{LOG_DIR}/gpu0.log').readlines()[-25:]))
        print('--- GPU 1 LOG TAIL ---\n' + ''.join(open(f'{LOG_DIR}/gpu1.log').readlines()[-25:]))
        raise RuntimeError(f'Worker error: p0={p0.returncode} p1={p1.returncode}')
else:
    !OMP_NUM_THREADS=2 {BLENDER_BIN} --background --python src/build_newton_scene.py -- --start-frame 1 --end-frame 720 --fps 24 --samples 12 --scale 75

rendered = len(glob.glob('render_frames/frame_*.jpg'))
print(f'Render complete: {rendered}/{TOTAL} frames in {(time.time()-t0)/60:.1f} min!')
```
